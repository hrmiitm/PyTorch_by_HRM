```python
import torch

if torch.cuda.is_available():
    torch.device("cuda")
    print(f"Using GPU: {torch.cuda.get_device_name(0)}")
else:
    torch.device("cpu")
    print(f"GPU not available. Using CPU: {torch.get_device}")
```

# CreatingTensors

```python
# tensor, empty
torch.tensor([[1,2,3], [0.5, 1, 2]]), torch.empty(2,3)
```

```python
# zeros, ones, rand
torch.manual_seed(100)
torch.zeros(2,3), torch.ones(2,3), torch.rand(2,3)
```

# ElementWise Operations

```python
# ElementWise MathsOperation
a = torch.tensor([[1,2,3], [4,5,6]])
b = torch.tensor([[2,3,4], [5,6,7]])

a+2
a-2
a*2
a/2 # converted to float
a//2
a%2
a**2

a+b
a-b
a*b
a/b
b//a
a**b
```

```python
# ElementWise More Operation
a = torch.tensor([[1,2,3], [-4,-5,-6]])

torch.abs(a)
torch.round(a)
torch.negative(a)
torch.ceil(a)
torch.floor(a)
torch.clamp(a, min=2, max=4) # IMP
```

# Matrix Operations

```python
a = torch.randint(size=(2,3), low=0, high=10).float()
b = torch.randint(size=(3,2), low=0, high=10).float()

a,b
```

```python
m = torch.matmul(a,b)
d = torch.det(m)
i = torch.inverse(m)
mm = torch.matmul(m,i)
m,d,i,mm
```

```python
v1 = torch.tensor([1,2])
v2 = torch.tensor([3,4])

torch.dot(v1,v2)
```

# InPlace Operations

```python
a = torch.randint(size=(2,3), low=-5, high=10).float()
b = torch.randint(size=(2,3), low=-5, high=10).float()

a,b
```

```python
a.add_(b)
```

```python
torch.relu(a), a
```

```python
torch.relu_(a), a
```

# Special Functions

```python
a = torch.randint(size=(2,3), low=-5, high=5).float()
a
```

```python
a.exp(), a.log(), a.sqrt() # a.log uses natural logrithm
```

```python
a.sigmoid(), a.softmax(dim=0), a.relu()
```
