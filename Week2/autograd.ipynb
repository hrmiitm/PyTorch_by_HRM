## Example1: 
- $y=x^{2}$

````python
data = torch.tensor(3.0, requires_grad=True)
y = data**2

data, y
````

````python
y.backward()
````

# Example3

```mermaid
    graph LR
        X -->|w| f(("wx+b | sigmoid"))
        f --> o[y_pred]
```

- $\frac{\partial \text{ Loss}}{\partial w}$

- $\frac{\partial \text{ Loss}}{\partial b}$

````python
# INPUT
x = torch.tensor(6.7)
y = torch.tensor(0.0)

# Parameters
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)
````

````python
def binary_cross_entropy(prediction, target):
    ep = 1e-8 # Epsilon to prevent log(0)
    prediction = torch.clamp(prediction, ep, 1-ep)

    a = target * torch.log(prediction)
    b = (1-target) * torch.log(1-prediction)
    return -1*(a+b)
````

````python
# Forward Pass
z = w*x + b
y_pred = torch.sigmoid(z)
````

````python
# Loss
loss = binary_cross_entropy(y_pred, y)
````

````python
# Backward Pass
loss.backward()
print(w.grad, b.grad) # (tensor(6.6918), tensor(0.9988))

w.grad.zero_(), b.grad.zero_()
````

- **requires_grad_(False):**  
  - Use to "freeze" model parameters so they are not updated during training (common in transfer learning or when you want to exclude some layers from gradient computation).

- **detach():**  
   - Use to break a tensor from the computation graph, stopping gradient tracking for that tensor-helpful for extracting intermediate results, logging, or custom training loops without affecting gradients.

- **torch.no_grad():**  
  - Use as a context manager to temporarily disable gradient tracking for all operations inside the block-ideal for inference/validation to reduce memory usage and speed up computations.
